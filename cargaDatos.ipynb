```python
from pathlib import Path
import pandas as pd
```

```python
file_path = Path("sp500_history.parquet")
file_id = "1nvubXdAu0EONIrP_yrURZbnPhBQ-uDaB"

url_primary = f"https://drive.usercontent.google.com/download?id={file_id}&export=download&confirm=t"
url_fallback = f"https://drive.google.com/uc?export=download&id={file_id}&confirm=t"

if not file_path.exists():
    print("Archivo no encontrado. Descargando desde Google Drive...")
    try:
        df = pd.read_parquet(url_primary)
    except Exception:
        df = pd.read_parquet(url_fallback)
    df.to_parquet(file_path)
    print("Descarga completada.")
else:
    df = pd.read_parquet(file_path)
```

```python
#Exporto el df como uno nuevo para no modificar el original
df.to_parquet("sp500_history_copy.parquet")
```
